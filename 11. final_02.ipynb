## Feature Engineering (신규 피처 7개)

7개 신규 피처는 서로 상관관계가 낮도록(짝지어 |corr| < 0.5) train.csv로 직접 계산해서 골랐다:
`deck`, `is_married_woman`, `embarked`(복원), `fare_per_person`, `age_group`, `ticket_prefix_grp`, `is_alone`.

`deck`은 `cabin`의 결측 여부 자체가 중요한 신호라서, 뒤에 나오는 일괄 결측치 대체(전체 컬럼을
평균/최빈값으로 채우는 셀)보다 **먼저** 원본 `cabin`에서 추출한다 (안 그러면 결측 여부 정보가
최빈값으로 덮여서 사라짐).

```python
# cabin 결측 여부가 곧 신호이므로, 아래 일괄 결측치 대체보다 먼저 추출
for df in (train, test):
    df['deck'] = df['cabin'].astype(str).str[0]
    df['deck'] = df['deck'].where(df['cabin'].notna(), 'U')
```

```python
# 결측치 확인 및 제거 
null_tr = train.isnull().sum()

(null_tr[null_tr > 0] / train.shape[0]).sort_values(ascending=False)
```

```python
print(f"before: {train.isnull().sum().sum()} / {test.isnull().sum().sum()}")

for tmp in [train, test]:
    null_tmp = tmp.isnull().sum()
    null_cols = null_tmp[null_tmp > 0].index

    for col in null_cols:
        try:
            tmp[col] = tmp[col].fillna(train[col].mean())

        except:
            tmp[col] = tmp[col].fillna(train[col].mode().values[0])

print(f"after: {train.isnull().sum().sum()} / {test.isnull().sum().sum()}")
```

```python
ticket_counts = pd.concat([train['ticket'], test['ticket']]).value_counts()

train['ticket_group_size'] = train['ticket'].map(ticket_counts)
test['ticket_group_size'] = test['ticket'].map(ticket_counts)
```

```python
import re

def _ticket_prefix(t):
    t = str(t).replace('.', '').replace('/', '').upper()
    m = re.match(r'^([A-Z]+)', t)
    return m.group(1) if m else 'NONE'

for df in (train, test):
    df['ticket_prefix'] = df['ticket'].apply(_ticket_prefix)

_prefix_counts = pd.concat([train['ticket_prefix'], test['ticket_prefix']]).value_counts()
for df in (train, test):
    df['ticket_prefix_grp'] = df['ticket_prefix'].where(
        df['ticket_prefix'].map(_prefix_counts) >= 15, 'OTHER'
    )
```

```python
import re
TITLE_MAP = {
    'Mlle': 'Miss', 'Ms': 'Miss', 'Mme': 'Mrs', 'Col': 'Military', 'Major': 'Military', 'Capt': 'Military',
    'Don': 'Noble', 'Dona': 'Noble', 'Sir': 'Noble', 'Lady': 'Noble', 
    'Jonkheer': 'Noble', 'the Countess': 'Noble',
}

def extract_title(name):
    m = re.search(r',\s*([^\.]*)\.', name)
    raw = m.group(1).strip() if m else 'Unknown'
    return TITLE_MAP.get(raw, raw)

for df in (train, test):
    df['title'] = df['name'].apply(extract_title)
```

나머지 5개는 `title`/`age`/`fare`/`sibsp`/`parch`/`embarked`가 준비된 뒤 바로 계산한다.

```python
for df in (train, test):
    # 결혼한 여성 여부 (title == 'Mrs') — 생존율과 상관이 가장 높았던 신규 피처
    df['is_married_woman'] = ((df['gender'] == 'female') & (df['title'] == 'Mrs')).astype(int)

    # 가족 단위가 아니라 '1인당' 요금 — 큰 가족이 표 하나로 묶여 있을 때 raw fare와 다른 정보를 준다
    df['fare_per_person'] = df['fare'] / (df['sibsp'] + df['parch'] + 1)

    # 혼자 탑승 여부
    df['is_alone'] = ((df['sibsp'] + df['parch']) == 0).astype(int)

    # 연령대 구간 (결측은 이미 평균으로 채워진 뒤라 'unknown' 버킷은 실질적으로 안 씀)
    df['age_group'] = pd.cut(
        df['age'], bins=[-1, 12, 19, 59, 200],
        labels=['child', 'teen', 'adult', 'senior']
    ).astype(str)

# embarked는 아래 일괄 결측치 대체 단계에서 이미 최빈값으로 채워졌음 (별도 처리 불필요)
```

```python
cat_cols = ['pclass', 'gender', 'title']
```

```python
train.drop(['name', 'ticket', 'ticket_prefix', 'cabin'], axis=1, inplace=True)
test.drop(['name', 'ticket', 'ticket_prefix', 'cabin'], axis=1, inplace=True)
```

```python
cat_cols = [
    'pclass', 'gender', 'title', 'deck', 'embarked', 'age_group', 'ticket_prefix_grp'
]
```

**버그 수정**: `deck`(`T` 데크는 단 1명), `ticket_prefix_grp` 같은 희귀 카테고리가 홀드아웃
분리 후 train/test 한쪽에만 나타나면 XGBoost가 "unseen category" 에러를 던지고
(`fit_evaluation`의 `except`에 조용히 먹혀서 후보에서 통째로 빠짐), LightGBM도
"categorical_feature mismatch"로 실패했다. `train`/`test` 합쳐서 카테고리 전체 집합을
고정해두면 이후 어떤 부분집합(홀드아웃, CV fold, 최종 test)에서도 카테고리가 안정적으로 유지된다.

```python
for col in cat_cols:
    categories = sorted(set(train[col].unique()) | set(test[col].unique()), key=str)
    train[col] = pd.Categorical(train[col], categories=categories)
    test[col] = pd.Categorical(test[col], categories=categories)
```

## Cross Validation

```python
from sklearn.model_selection import KFold, StratifiedKFold
from sklearn.metrics import accuracy_score

def do_cross_validation(
    model: object,
    train: pd.DataFrame,
    target_col_name: str,
    eval_fnc: object = accuracy_score,
    is_classifier: bool = False,
    n_splits: int = 5,
    shuffle: bool = True,
    random_state: int = 42,
    use_proba: bool = False,
    **kargs
):
    """
    Cross Validation 객체 생성
    """
    if is_classifier:
        skf = StratifiedKFold(
            n_splits=n_splits, shuffle=shuffle, random_state=random_state, **kargs
        )
        gen_cv = skf.split(train, train[target_col_name])
    else:
        kf = KFold(
            n_splits=n_splits, shuffle=shuffle, random_state=random_state, **kargs
        )
        gen_cv = kf.split(train)

    """
    Cross Validation을 이용한 모델 학습
    """
    lst_scores = []
    # 학습 시작
    for tr_index, val_index in gen_cv:
        # 학습용, 검증용 데이터 생성
        x_tr, y_tr = train.iloc[tr_index].drop([target_col_name], axis=1), train.iloc[tr_index][target_col_name]
        x_te, y_te = train.iloc[val_index].drop([target_col_name], axis=1), train.iloc[val_index][target_col_name]

        # 모델 학습
        model.fit(x_tr, y_tr)
        # 예측
        if use_proba:
            pred = model.predict_proba(x_te)[:, 1]
        else:
            pred = model.predict(x_te)
        # 평가
        lst_scores.append(eval_fnc(y_te, pred))

    # 학습완료
    print(f"교차 검증 점수({eval_fnc.__name__}): {np.mean(lst_scores)}")
    return lst_scores
```

```python
from catboost import CatBoostClassifier

model = CatBoostClassifier(verbose=0, cat_features=cat_cols, random_state=42)

cv_scores_acc = do_cross_validation(
    model=model,
    train=train,
    target_col_name='survived',
    is_classifier=True,   # 생존 비율(62:38) 유지하며 fold 나누도록
    n_splits=5,
)
```

```python
from sklearn.metrics import roc_auc_score

cv_scores_auc = do_cross_validation(
    model=model,
    train=train,
    target_col_name='survived',
    eval_fnc=roc_auc_score,
    is_classifier=True,
    n_splits=5,
    use_proba=True,
)
```
